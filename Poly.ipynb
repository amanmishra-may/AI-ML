```python
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures

data = {
    "Hours" : [2,4,6,8,10],
    "Sleep" : [8,6,7,5,9]
}

df = pd.DataFrame(data)

poly = PolynomialFeatures(degree=2)

new_data = poly.fit_transform(df)

new_df = pd.DataFrame(
    new_data,
    columns= poly.get_feature_names_out(df.columns)
)
print(new_df)
```
