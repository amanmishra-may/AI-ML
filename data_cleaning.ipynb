```python
import pandas as pd
import numpy as np

data = {
    'Name':   ['Alice', 'Bob', 'invalid data', 'David', 'Eva', '###', 'Grace'],
    'Gender': ['F', 'M', 'M', 'invalid data', 'F', 'M', 'banana'],
    'Age':    [25, 30, -5, 22, 28, 200, 27]
}

df = pd.DataFrame(data)

ng = df.copy()

invalid_string = ng.isin(["invalid data"]).any(axis=1)
invalid_age = (ng["Age"] > 100) | (ng["Age"] < 0)
invalid_gender = ~ng["Gender"].isin(["M", "F"])

invalid_data = invalid_age | invalid_gender | invalid_string

clean = ng[~invalid_data] # inside quotes means data from that particular column and '~'  it means opposite

print(clean)
```
