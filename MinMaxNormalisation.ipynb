```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

data = {
    "Height": [160, 170, 180, 190],
    "Weight": [50, 65, 80, 95]
}

df = pd.DataFrame(data)
print("Before\n")
print(df)

scaler = MinMaxScaler()

scaler.fit(df)
                                        
scaled_data = scaler.transform(df)

df_scaled = pd.DataFrame(
    scaled_data,
    columns=df.columns
)

print("After\n")
print(df_scaled)
```
