```python
# Practical 11 — Text Preprocessing -> Stemming Technique
import pandas as pd
from nltk.stem import PorterStemmer

data = {
    "text" : [
        "I am playing cricket",
        "She studies daily",
        "They are playing games"
    ]
}

df = pd.DataFrame(data)
stemmer = PorterStemmer()

df["Stemed_data"] = df["text"].apply(
    lambda text :[
        stemmer.stem(word)
       for word in text.split()
    ]
)

print(df)
```
