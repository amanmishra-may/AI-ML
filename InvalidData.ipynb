```python
import pandas as pd

data = {
    "Name": ["Aman", "Rahul", "Ravi", "12345", "Priya"],
    "Gender": ["Male", "Male", "Male", "Male", "Female"],
    "Age": [20, 21, 19, 18, 20]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Identify invalid names
invalid_names = df[~df["Name"].str.isalpha()]

print("\nInvalid Entries:")
print(invalid_names)

# Remove invalid names
df = df[df["Name"].str.isalpha()]

print("\nClean Dataset:")
print(df)
```
