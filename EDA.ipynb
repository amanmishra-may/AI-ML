```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Age" :[18, 19, 19, 20, 20, 20, 21, 22, 22, 23, 25]
}

df = pd.DataFrame(data)

plt.hist(df["Age"],bins=5)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Distribution of Age")
plt.show()



plt.boxplot(df["Age"])
plt.ylabel("Age")
plt.title("Box Plot")
plt.show()
```
